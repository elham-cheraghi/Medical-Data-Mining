```python
import pandas as pd
from sklearn.metrics import confusion_matrix

d = pd.read_csv("CV_predictions.csv")
y = d["observed_outcome"]
p = (d["predicted_probability"] >= 0.3).astype(int)

tn, fp, fn, tp = confusion_matrix(y, p).ravel()

print(f"Confusion matrix:\ntn: {tn}\nfp: {fp}\nfn: {fn}\ntp: {tp}")
print("Sensitivity:", tp / (tp + fn))
print("Specificity:", tn / (tn + fp))
print("PPV:", tp / (tp + fp))
print("NPV:", tn / (tn + fn))
print("Accuracy:", (tp + tn) / len(d))
```

```python
import pandas as pd
from sklearn.metrics import confusion_matrix

d = pd.read_csv("CV_predictions.csv")
rows = []

for fold, g in d.groupby("fold"):
    y = g["observed_outcome"]
    p = (g["predicted_probability"] >= 0.3).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, p, labels=[0, 1]).ravel()

    rows.append({
        "Fold": fold,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
    })

results = pd.DataFrame(rows)
results
```
