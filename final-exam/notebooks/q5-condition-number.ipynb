```python
import pandas as pd
import numpy as np

data = pd.read_csv('outlier_data.csv')
data_var = data[["variable_1", "variable_2", "variable_3"]]

correlation_matrix = data_var.corr()

data_X = np.column_stack([
    np.ones(data_var.shape[0]),
    data_var.to_numpy(),
])
cond_raw = np.linalg.cond(data_X)

z = (data_var - data_var.mean()) / data_var.std(ddof=1)
data_X_z = np.column_stack([
    np.ones(z.shape[0]),
    z,
])
cond_z = np.linalg.cond(data_X_z)

print(f"Correlation matrix:\n{correlation_matrix}")
print(f"Condition number of the raw data: {cond_raw:.4f}")
print(f"Condition number of the standardized data: {cond_z:.4f}")
```
