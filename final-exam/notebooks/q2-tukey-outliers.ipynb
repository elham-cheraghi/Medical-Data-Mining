```python
import pandas as pd
import numpy as np

data = pd.read_csv("outlier_data.csv")
data_v3 = data["variable_3"]

mean_v3 = data_v3.mean()
median_v3 = data_v3.median()
std_v3 = data_v3.std(ddof=1)
pearson_skewness = 3 * (mean_v3 - median_v3) / std_v3
print(f"Pearson skewness = {pearson_skewness:.3f}")


Q1 = np.percentile(data_v3, 25, method="weibull")
Q3 = np.percentile(data_v3, 75, method="weibull")
IQR = Q3 - Q1

tukey_low = round(Q1 - 1.5 * IQR, 3)
tukey_up = round(Q3 + 1.5 * IQR, 3)

outliers_mask = (data_v3 < tukey_low) | (data_v3 > tukey_up)
outliers = data[outliers_mask][["sample_id", "variable_3"]].round(3)

print(f"Q1 = {Q1:.3f}")
print(f"Q3 = {Q3:.3f}")
print(f"IQR = {IQR:.3f}")
print(f"Tukey lower and upper bounds:\n[{tukey_low:.3f}, {tukey_up:.3f}]")
outliers
```
